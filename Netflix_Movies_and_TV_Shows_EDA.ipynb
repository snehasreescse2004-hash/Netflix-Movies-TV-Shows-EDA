```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
missing_data = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': missing_percentage
})

print(missing_data)
```

```python
missing_data = missing_data[missing_data['Missing Values'] > 0]

plt.figure(figsize=(10, 5))

plt.bar(missing_data.index, missing_data['Missing Values'])

plt.title("Missing Values in Netflix Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

plt.show()
```

**Observation**

From the above analysis, I found that the director column has the highest number of missing values (2634). The cast and country columns also contain some missing values.

The date_added, rating and duration columns have very few missing values. The bar chart clearly shows that most of the missing values are present in the director column.

```python
#Movies vs TV Shows
content_type = df['type'].value_counts()
plt.figure(figsize=(7, 5))
plt.bar(content_type.index, content_type.values)
plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

### Observation

From the above chart, I can compare the number of Movies and TV Shows available on Netflix. The chart helps me understand which type of content is more common on the platform.

```python
#Top 10 Countries
top_countries = df['country'].value_counts().head(10)
plt.figure(figsize=(10, 5))
plt.bar(top_countries.index, top_countries.values)
plt.title("Top 10 Countries on Netflix")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

### Observation

From the above chart, I found that Dramas, International Movies have the highest number of titles among the displayed categories. Documentaries and Stand-Up Comedy are also popular categories on Netflix.

I observed that different combinations of genres are available on Netflix. This chart helped me understand which genre combinations have more titles on the platform.

```python
#Content Released Each Year
year_count = df['release_year'].value_counts().sort_index()
plt.figure(figsize=(12, 5))
plt.plot(year_count.index, year_count.values)
plt.title("Netflix Content Released Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

### Observation

From the above chart, I observed that the number of Netflix titles was very low in the earlier years. The number of titles started increasing gradually after 2000 and increased rapidly after 2010.

The highest number of titles can be seen around 2018. After that, the number of titles decreased in the following years. This chart helped me understand how the release of content available on Netflix has changed over time.

```python
# Top 10 Genres

top_genres = df['listed_in'].value_counts().head(10)
plt.figure(figsize=(10, 5))
plt.bar(top_genres.index, top_genres.values)
plt.title("Top 10 Genres on Netflix")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45, ha='right')
plt.show()
```

### Observation

From the above chart, I found that Dramas, International Movies have the highest number of titles among the top 10 categories. Documentaries and Stand-Up Comedy are also among the popular categories on Netflix.

I observed that Netflix contains different combinations of genres, and some combinations have more titles than others. This chart helped me understand the distribution of different genre categories on Netflix.

```python
#Movie Duration Distribution
movies = df[df['type'] == 'Movie'].copy()

# Extracting movie duration in minutes
movies['duration'] = movies['duration'].str.replace(' min', '', regex=False)
movies['duration'] = pd.to_numeric(movies['duration'], errors='coerce')
plt.figure(figsize=(9, 5))
plt.hist(movies['duration'].dropna(), bins=20)
plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")
plt.show()
```

### Observation

From the above histogram, I observed that most of the movies on Netflix have a duration between 80 and 110 minutes. The number of movies is higher in this duration range compared to shorter and longer movies.

I also noticed that only a few movies have a duration above 200 minutes. This chart helped me understand the distribution of movie durations available on Netflix.

```python
# Content Ratings
rating_count = df['rating'].value_counts().head(10)
plt.figure(figsize=(9, 5))
plt.bar(rating_count.index, rating_count.values)
plt.title("Top 10 Content Ratings on Netflix")
plt.xlabel("Content Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

### Observation

From the above chart, I found that TV-MA has the highest number of titles on Netflix, followed by TV-14. TV-PG and R are also among the common content ratings.

I observed that Netflix has content available for different age groups. This chart helped me understand the distribution of content ratings on the platform.

```python
#Movies vs TV Shows by Release Year
content_year = df.groupby(['release_year', 'type']).size().unstack()
content_year.plot(figsize=(12, 5))
plt.title("Movies and TV Shows Released Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

### Observation

From the above chart, I observed that the number of Movies is higher than TV Shows in most of the release years. Both Movies and TV Shows started increasing gradually after 2000 and showed a rapid increase after 2010.

Movies reached their highest point around 2018, while TV Shows reached their highest point around 2020. This chart helped me compare the release patterns of Movies and TV Shows on Netflix.
